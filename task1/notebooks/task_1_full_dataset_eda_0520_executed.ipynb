## 2. Shape Of The Data

`source_summary` aggregates every source and symbol across the full analyzed date range. For trades and liquidations it includes notionals and side balance. For BBO it includes spread statistics.

```python
source_daily = read_table("source_daily")
source_daily["date"] = pd.to_datetime(source_daily["date"])
fig, axes = plt.subplots(4, 2, figsize=(14, 12), sharex=True)
axes = axes.flatten()
for ax, ((source, symbol), g) in zip(axes, source_daily.groupby(["source", "symbol"])):
    ax.plot(g["date"], g["rows"], linewidth=1.2)
    ax.set_title(f"{source} / {symbol}")
    ax.set_ylabel("rows/day")
    ax.tick_params(axis="x", rotation=30)
savefig("02_rows_per_day_all_sources.png")
```

```python
source_hourly = read_table("source_hourly")
hourly_pattern = (
    source_hourly.groupby(["source", "symbol", "hour"], as_index=False)
    .agg(rows_mean=("rows", "mean"), notional_mean=("notional_sum", "mean"))
)
fig, axes = plt.subplots(4, 2, figsize=(14, 11), sharex=True)
axes = axes.flatten()
for ax, ((source, symbol), g) in zip(axes, hourly_pattern.groupby(["source", "symbol"])):
    ax.plot(g["hour"], g["rows_mean"], marker="o", linewidth=1)
    ax.set_title(f"Average hourly event rate: {source} / {symbol}")
    ax.set_xlabel("UTC hour")
    ax.set_ylabel("rows/hour")
savefig("03_hourly_event_pattern.png")
```

**Shape observations.**

## 5. Data Quality: Gaps, Duplicates, Broken Rows

The table below comes from full parquet scans. For the huge trade files, gap checks are calculated day by day to avoid memory blowups. Same-timestamp trades are common and expected in aggregated exchange trade feeds; they are not necessarily exact duplicates.

```python
gaps = read_table("timestamp_gaps")
gap_summary = (
    gaps.groupby(["source", "symbol"], as_index=False)
    .agg(
        rows=("rows", "sum"),
        negative_time_jumps=("negative_time_jumps", "sum"),
        same_timestamp_as_previous=("same_timestamp_as_previous", "sum"),
        gaps_gt_1s=("gaps_gt_1s", "sum"),
        gaps_gt_10s=("gaps_gt_10s", "sum"),
        gaps_gt_60s=("gaps_gt_60s", "sum"),
        dt_us_p50=("dt_us_p50", "median"),
        dt_us_p99=("dt_us_p99", "median"),
        dt_us_max=("dt_us_max", "max"),
    )
)
gap_summary.sort_values(["source", "symbol"])
```

```python
quality_cols = [
    "source", "symbol", "bad_price_rows", "bad_amount_rows", "null_rows",
    "crossed_rows", "negative_spread_rows", "bad_bid_amount_rows", "bad_ask_amount_rows",
]
existing = [c for c in quality_cols if c in source_daily.columns]
quality = source_daily[existing].fillna(0).groupby(["source", "symbol"], as_index=False).sum(numeric_only=True)
quality
```

**Quality observations.**

- No negative time jumps were found in the scans.
- Trade feeds have many repeated timestamps, which is expected at microsecond resolution and high event rates.
- BBO feeds have occasional larger gaps; this matters for markout construction and should be handled with forward-filled mid at the target timestamp.
- No obvious broken price/amount rows or crossed BBO rows dominate the dataset.

## 6. Conventions That Bite

The task description warns about three things: timestamp units, side conventions, and Bybit delay. I verified them on concrete examples and through aggregate checks.

- Timestamps are microseconds since UNIX epoch UTC.
- In Binance trades, `side=buy` means taker bought, so the hypothetical maker sold.
- In liquidation tables, `side=buy` means a forced buy order, i.e. short close pressure upward.
- Bybit liquidations are shifted by +200ms before cross-source relationship calculations.

## 8. Liquidations Around The Order Book And Trade Flow

For every liquidation event, I measured Binance mid returns before and after the liquidation over 5s, 30s, 120s, and 300s horizons. I also measured Binance trade flow before and after the event. This is computed over all liquidation events, not a sample.

```python
event_summary = read_table("liquidation_event_context_summary")
cols = [
    "symbol", "source", "side", "size_bucket", "events", "notional_sum",
    "mid_ret_before_30s_bps", "mid_ret_after_30s_bps", "signed_mid_ret_after_30s_bps",
    "mid_ret_after_120s_bps", "signed_mid_ret_after_120s_bps",
    "mid_ret_after_300s_bps", "signed_mid_ret_after_300s_bps",
]
event_summary[cols].sort_values(["symbol", "source", "side", "size_bucket"]).head(40)
```

```python
plot_df = event_summary.copy()
fig, axes = plt.subplots(1, 3, figsize=(16, 4), sharey=True)
for ax, horizon in zip(axes, ["30s", "120s", "300s"]):
    col = f"signed_mid_ret_after_{horizon}_bps"
    sns.barplot(data=plot_df, x="size_bucket", y=col, hue="source", ax=ax, errorbar=None)
    ax.axhline(0, color="black", linewidth=1)
    ax.set_title(f"Post-liquidation signed return, {horizon}")
    ax.tick_params(axis="x", rotation=30)
savefig("07_liquidation_signed_returns_by_size.png")
```

**Liquidation event observations.**

- Liquidations occur after strong same-direction price movement, visible in the pre-event returns.
- The post-event behavior is not simply continuation everywhere. Some large buckets show reversal or exhaustion, especially at longer horizons.
- Bybit liquidation events are often associated with larger pre-event Binance price moves and wider BBO spread at the event time. This makes Bybit liquidations interesting as a cross-exchange stress signal.

## 9. Cross-Exchange Alignment

Binance and Bybit liquidation feeds live on the same time axis after shifting Bybit forward by +200ms. The table below measures how much liquidation notional is observed on the other exchange before and after events.

```python
align = read_table("liquidation_cross_exchange_alignment_daily")
align_summary = (
    align.groupby(["symbol", "source", "side", "size_bucket"], as_index=False)
    .agg(
        events=("events", "sum"),
        other_liq_notional_prev_1s=("other_liq_notional_prev_1s", "mean"),
        other_liq_notional_next_1s=("other_liq_notional_next_1s", "mean"),
        other_liq_notional_prev_5s=("other_liq_notional_prev_5s", "mean"),
        other_liq_notional_next_5s=("other_liq_notional_next_5s", "mean"),
        other_liq_notional_prev_30s=("other_liq_notional_prev_30s", "mean"),
        other_liq_notional_next_30s=("other_liq_notional_next_30s", "mean"),
    )
)
align_summary.head(40)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4), sharey=True)
for ax, window in zip(axes, ["1s", "5s"]):
    tmp = align_summary.copy()
    tmp["next_minus_prev"] = tmp[f"other_liq_notional_next_{window}"] - tmp[f"other_liq_notional_prev_{window}"]
    sns.barplot(data=tmp, x="size_bucket", y="next_minus_prev", hue="source", ax=ax, errorbar=None)
    ax.axhline(0, color="black", linewidth=1)
    ax.set_title(f"Other exchange liq after-before, {window}")
    ax.tick_params(axis="x", rotation=30)
savefig("08_cross_exchange_alignment.png")
```

## 10. Weird / Important Things Found

- The dataset is large enough that naive pandas-first EDA is the wrong tool. Full exploration needs parquet-aware scans and compact aggregate tables.
- Trade timestamps repeat frequently; this is normal for high-frequency exchange data and should not be treated as automatic duplication.
- BBO updates are much less frequent than trades, so trade-to-BBO joins must be as-of joins.
- Liquidations are sparse but heavy-tailed and clustered.
- The most dangerous convention mistake is sign: taker buy means maker sell.
- Bybit must be shifted by +200ms before using it as a predictor for Binance-time decisions.

## 11. Limitations And Data Caveats

- The BBO data contains only top-of-book information, not full depth or queue position.
- A hypothetical maker-fill analysis is an approximation: real maker execution depends on queue position, order placement, cancels, and latency.
- Same-timestamp trades are common and should not be treated as automatic duplicates.
- Bybit liquidations must be sorted and delayed by +200ms before being used as Binance-time features.
- Large liquidation events are rare, so bucket-level estimates for extreme sizes can be noisy.
- Hidden test dates may have different market regimes, so findings should be validated temporally, not with random splits.

## 13. Final Takeaways

This EDA answers the starting questions from the assignment on the full analyzed dataset:

- **Shape:** event rates differ strongly by source and symbol; trades are dense, BBO is frequent but lower-frequency than trades, and liquidations are sparse and bursty.
- **Distributions:** trade and liquidation notionals are heavy-tailed. Liquidation tails are large enough that p99/p999/max are more useful than mean alone.
- **Relationships:** trade-to-BBO alignment confirms the trade side convention; liquidation event studies show stressed regimes with strong pre-event movement and mixed continuation/exhaustion after the event; Bybit has visible cross-exchange alignment with Binance after the +200ms delay.
- **Conventions:** timestamps are UTC microseconds, Binance trade side is taker side, liquidation side is forced order side, and Bybit must be shifted forward by +200ms.
- **Weird things:** repeated trade timestamps are common and expected; Bybit liquidation raw files contain some backward timestamp jumps, so sorting before joins is mandatory; BBO has occasional larger gaps, so markouts must use careful as-of logic.

The main research implication is that liquidation flow should not be treated as a naive one-direction trend signal. It is better interpreted as a stressed microstructure regime where maker fill quality depends on side, event size, source, and horizon.
